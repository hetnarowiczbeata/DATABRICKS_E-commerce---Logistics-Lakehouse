```python
%sql
USE CATALOG ecommerce_logistics;
USE SCHEMA feedback
```

```python
%sql
DESCRIBE TABLE bronze_feedback
```

```python
df=spark.table("bronze_feedback")
```

```python
from pyspark.sql.functions import col,to_timestamp,expr
```

```python
# df.select("review_score").distinct().show()
df.select("review_answer_timestamp").distinct().show()
```

```python
silver_feedback=spark.table("bronze_feedback").withColumn("review_score",expr("try_cast(review_score as int)")).withColumn("review_creation_date",to_timestamp(col("review_creation_date"))).withColumn("review_answer_timestamp",to_timestamp(col("review_answer_timestamp"))).filter(col("review_score").between(1,5)).filter(col("review_creation_date").isNotNull()).filter(col("review_answer_timestamp").isNotNull())
```

```python
silver_feedback.write.mode("overwrite").saveAsTable("silver_feedback")
```
