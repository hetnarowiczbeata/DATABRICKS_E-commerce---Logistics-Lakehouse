```python
%sql
USE CATALOG ecommerce_logistics;
USE SCHEMA products
```

```python
silver_products=spark.table("bronze_products")
```

```python
silver_products.groupBy("product_category_name").count().orderBy("count", ascending=False).show(100,truncate=False)
```

```python
from pyspark.sql.functions import col
```

```python
silver_products.filter(col("product_category_name").isNull()).show(150)
```

```python
silver_products=silver_products.withColumnRenamed("product_name_lenght", "product_name_length").withColumnRenamed("product_description_lenght", "product_description_length").filter(col("product_category_name").isNotNull()).write.format("delta").mode("overwrite").saveAsTable("silver_products")
```
