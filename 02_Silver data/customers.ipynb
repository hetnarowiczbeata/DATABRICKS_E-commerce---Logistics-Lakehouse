```python
%sql
USE CATALOG ecommerce_logistics;
USE SCHEMA customers
```

```python
%sql
DESCRIBE TABLE bronze_customers
```

```python
customers_bronze=spark.table("bronze_customers")
```

```python
# customers_bronze.select("customer_state").distinct().show()
customers_bronze.select("customer_city").distinct().show()

```

```python
from pyspark.sql.functions import col,trim,upper

```

```python
customers_silver=(customers_bronze.withColumn("customer_zip_code_prefix",col("customer_zip_code_prefix").cast("string")).withColumn("customer_city",trim(col("customer_city"))))
```

```python
customers_silver.write.format("delta").saveAsTable("silver_customers")
```
