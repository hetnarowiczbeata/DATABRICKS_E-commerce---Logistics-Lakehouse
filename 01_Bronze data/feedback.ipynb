```python
%sql
USE CATALOG ecommerce_logistics;
USE SCHEMA feedback
```

```python
df_feedback = (
    spark.read
    .format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .option("quote", '"')
    .option("escape", '"')
    .option("multiLine", True)
    .load("/Volumes/ecommerce_logistics/feedback/raw_data/olist_order_reviews_dataset.csv")
)

df_feedback.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("feedback.bronze_feedback")
```
